```python
import pandas as pd

events = pd.DataFrame({"Day": ["Fri", "Fri", "Fri", "Fri", "Fri", "Fri", "Fri", "Fri", "Fri", "Fri","Sat", "Sat", "Sat", "Sat", "Sat", "Sat", "Sat", "Sat", "Sat", "Sat"], "Tickets": [52, 47, 61, 59, 55, 63, 58, 49, 57, 62, 88, 91, 95, 84, 99, 90, 93, 105, 87, 89]})
```

```python
import matplotlib.pyplot as plt
```

```python
events.boxplot(column="Tickets", by="Day")
plt.title ("Tickets Sold on Friday vs Saturday")
plt.suptitle("")  
plt.xlabel("Day of Event")
plt.ylabel("Tickets Sold")

plt.show()  
```

```python
print("The boxplot shows that the median number of tickets sold on Saturdays is much higher than on Fridays, suggesting that Saturdays attract more attendees due to more leisure time perhaps or events are able to have greater accessiblity to headliners, marketing opportunities, etc. The Saturday boxplot also shows a lower spread between Q1 and Q3 compared to Friday, which shows less varience in ticket sales and lower business risk. Additionally, for the Saturday boxplot, there is an outlier point which may infer that for this single event there was potentially a famous DJ playing or that weekend was a bank holiday.")
```

```python
import pandas as pd

values = pd.DataFrame({"Weekday or Weekend": ["Weekend", "Weekday", "Weekday", "Weekday", "Weekday", "Weekday", "Weekend", "Weekend", "Weekday", "Weekday", "Weekday", "Weekday", "Weekday", "Weekend", "Weekend", "Weekday", "Weekday"], "Event Views": [12,16,5,8,2,4,14,35,10,6,22,10,10,2,3,41,9]})
```

```python
values.boxplot(column="Event Views", by="Weekday or Weekend")

plt.title("Event Views on Weekdays vs Weekends")
plt.suptitle("")    
plt.xlabel("Weekday or Weekend")
plt.ylabel("Event Views")

plt.show()
```

```python
print("Both boxplots show a median which is close in magnitude with the weekend boxplot median value being slightly higher. Weekend boxplot shows a lower overall variance between the data points, showing a more consistent marketing performance perhaps. Both boxplots show outlier points which are significantly larger than compared to the rest of the data, highlighting that such points could infer an important announcement occured on that day.")
```
