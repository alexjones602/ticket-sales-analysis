```python
import pandas as pd
df = pd.read_csv("../data/JE001_real_sample.csv")
```

```python
print("How have ticket sales changed over time for Jigsaw Events Volume 001?")
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df["date"], df["tickets_sold"])

plt.title("House Music Event Ticket Sales Over Time")
plt.xlabel("Date")
plt.ylabel("Tickets Sold")

plt.xticks(rotation=45)
plt.grid()  

plt.savefig("../outputs/JE001_ticket_sales_over_time.png")

plt.show()
```

```python
print("The graph shows that ticket sales generally increase over time, with sharper increases such as the one in early August. Such trend cannot be explained by the relationship between dates and ticket sales, but also factors such as significant marketing shifts, increased following, consumer confidence etc.  ")
```

```python
print("How does Instagram engagement relate to ticket sales for Jigsaw Events Volume 001?")
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(df["instagram_followers"], df["tickets_sold"])

plt.title("Instagram Engagement vs Ticket Sales")
plt.xlabel("Instagram Followers")
plt.ylabel("Tickets Sold")

plt.savefig("../outputs/JE001_ticket_sales_instagram_followers.png")

plt.grid()
plt.show()
```

```python
plt.savefig("../outputs/JE001_ticket_sales_instagram_followers.png")
```

```python
print("The graph suggests that there is a positive, but weak correlation between Instagram followers and ticket sales. The graph shows a spread of points (no real correlation). However, generally as instagram followers increase, ticket sales also increase. This infers that as individuals see the following increasing, they may feel they will 'miss-out' and purchase a ticket as others follow the account and have purchased their ticket already. This is a common marketing tactic used to increase sales, and is known as 'social proof'.")
```

```python
import pandas as pd
df = pd.read_csv("../data/JE001_real_advertisement.csv")
```

```python
df.head(20)
```

```python
print("How do Weekdays and the Weekend differ when it comes to respective event engagment levels (event views and ticket sales)?")
```

```python
df["weekend/weekday"] = df["day"].apply(lambda x: "Weekend" if x in ["Saturday", "Sunday"] else "Weekday")
```

```python
df.head(20)
```

```python
df["event_engagement"]= df["tickets_sold"] + df["event_views"]
```

```python
import matplotlib.pyplot as plt

df.boxplot(column="event_engagement",by="weekend/weekday")

plt.title("Event Engagement: Weekday vs Weekend")
plt.suptitle("")
plt.xlabel("Day Type")
plt.ylabel("Event Engagement")

plt.savefig("../outputs/JE001_weekend_vs_weekday_engagement.png", dpi=300,bbox_inches="tight")

plt.show()
```

```python
print("When comparing the two boxplots, the Weekend has a higher median infering a higher level of engagement. Additionally, the Weekend shows a higher IQR indicating a higher variation in a half of the values regarding event engagement. However the Weekday shows a greater spread hinting greater overall variation in event engagement compared to the Weekend. Both boxplots have outlier points that are isolated compared to the rest of their respective data sets, infering that external factors such as important announcements caused event engagment to increase significantly on such days. Overall, the results show that the Weekend tends to generate higher engagement while being less variant as a whole. Highlighting that individuals are more responsive to announcements and the event in general on weekends.")
```
