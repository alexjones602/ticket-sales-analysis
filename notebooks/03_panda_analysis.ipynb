```python
print("Which entries / days sold over 2 tickets?")
```

```python
over_2_tickets = df[df["tickets_sold"] > 2]
over_2_tickets.sort_values("tickets_sold", ascending=False)
```

```python
print("What was the highest number of tickets sold on a single day?")   
```

```python
highest_number_of_tickets_sold = df.sort_values("tickets_sold",ascending=False)
highest_number_of_tickets_sold[["day", "tickets_sold", "event_views"]].head(1)
```
