```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../outputs/cleaned_event_data.csv")
```

```python
df.head(11)
```

```python
df["ticket_price"]
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["ticket_price"], bins=4, edgecolor="black")

plt.title("Distribution of Ticket Prices")

plt.xlabel("Ticket Price (£)")

plt.ylabel("Number of Events")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["tickets_sold"], bins=5, edgecolor="black")

plt.title("Distribution of Tickets Sold")

plt.xlabel("Tickets Sold")

plt.ylabel("Number of Events")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["days_advertised"], bins=5, edgecolor="black")

plt.title("Distribution of Days Advertised")

plt.xlabel("Days Advertised")

plt.ylabel("Number of Events")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["capacity"], bins=4, edgecolor="black")

plt.title("Distribution of Event Capacities")

plt.xlabel("Event Capacity")

plt.ylabel("Number of Events")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/JE001_real_advertisement.csv")
```

```python
df.head(17)
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["event_views"], bins=5, edgecolor="black")

plt.title("Distribution of Event Views")

plt.xlabel("Event Views")

plt.ylabel("Number of Events")

plt.show()
```

```python
df["event_interactions"] = df["tickets_sold"] + df["event_views"]
```

```python
plt.figure(figsize=(8,6))

plt.hist(df["event_interactions"], bins=5, edgecolor="black")       

plt.title("Distribution of Event Interactions") 

plt.xlabel("Event Interactions")

plt.ylabel("Number of Events")

plt.show()
```
