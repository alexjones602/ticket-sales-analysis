```python
average_price = df["ticket_price"].mean()

average_price
```

```python
df["ticket_price"] = df["ticket_price"].fillna(average_price)
```

```python
df["ticket_price"].isna().sum()
```

```python
df["weather"] = df["weather"].fillna("Rain")
```

```python
df["weather"].isna().sum()
```

```python
print("4.2 Identifying impossible values")
```

```python
df.describe()
```

```python
df[df["capacity"] <= 0]
```

```python
df.loc[df["event_name"] == "Groovy Theory", "capacity"] = 180
```

```python
df[df["tickets_sold"] > df["capacity"]]
```

```python
df[df["ticket_price"] <= 0]
```

```python
df[df["tickets_sold"] < 0]
```

```python
df[df["days_advertised"] < 0]
```

```python
print("4.3 Identifying incorrect data types")
```

```python
df.dtypes
```

```python
df["ticket_price"] = pd.to_numeric(df["ticket_price"])
```

```python
df["ticket_price"] = pd.to_numeric(df["ticket_price"],errors="coerce")
```

```python
df["event_date"] = pd.to_datetime(df["event_date"],errors="coerce",dayfirst=True)
```

```python
df.dtypes
```

```python
df[df["event_date"].isna()]
```

```python
df.duplicated()
```

```python
df[df.duplicated(keep=False)]
```

```python
df = df.drop_duplicates()
```

```python
print("4.4 Final Report on data cleaning")
```

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))
print("Missing values:", df.isna().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Invalid capacities:", (df["capacity"] <= 0).sum())
print("Negative prices:", (df["ticket_price"] < 0).sum())
print("Negative tickets sold:", (df["tickets_sold"] < 0).sum())
print("Negative days advertised:", (df["days_advertised"] < 0).sum())
```
