```python
print("11.2 Do reel views affect ticket sales?")
```

```python
import pandas as pd
df = pd.read_csv("../data/JE001_real_sample.csv")
print(df.head(9))
```

```python
print(df.info())
print(df.describe())
```

```python
print(df[["reel_views", "tickets_sold"]].describe())
```

```python
correlation = df["reel_views"].corr(df["tickets_sold"])
print(correlation)
```

```python
import matplotlib.pyplot as plt
plt.scatter(df["reel_views"], df["tickets_sold"])
plt.xlabel("Reel Views")
plt.ylabel("Tickets Sold")
plt.title("Reel Views vs Tickets Sold")
plt.show()
```

```python
correlation = df["reel_views"].corr(df["instagram_followers"])
print(correlation)
```
